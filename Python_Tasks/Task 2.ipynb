### 1.Find the Shortest Word in a Sentence

### 2.Find the Largest Number in a List

```python
l=[12, 45, 7, 89, 23]
for i in range(0,len(l)):
    for j in range(i,len(l)):
        if l[i]>l[j]:
            l[i],l[j]=l[j],l[i]
print(l[-1])
```

### 3.Find the Smallest Number in a List

```python
l=[12, 45, 7, 89, 23]
for i in range(0,len(l)):
    for j in range(i,len(l)):
        if l[i]>l[j]:
            l[i],l[j]=l[j],l[i]
print(l[0])
```

### 6.Count Vowels in a String

```python
w='Education'
c=0
for i in w:
    if i in 'AEIOUaeiou':
        c+=1
print(c)
```

### 7.Count Consonants in a String

```python
w='Python 123!'
c=0
for i in w:
    if i.isalpha() and i not in 'AEIOUaeiou':
        c+=1
print(c)
```

### 8. Find the Most Frequent Character

```python
w='banana'
f={}
for i in w:
    f[i]=w.count(i)
print(max(f,key=f.get))
```

### 9. Find the Least Frequent Character

### 15. Create a Dictionary of Numbers and Their Even/Odd Status

```python
l=[1, 2, 3, 4, 5]
f={}
for i in l:
    if i%2==0:
        f[i]='Even'
    else:
        f[i]='Odd'
f
```

### 16. Find the Most Frequent Word in a Sentence

```python
s='Python is easy and Python is powerful'
f={}
for i in s.casefold().split():
    f[i]=s.casefold().count(i)
print(max(f,key=f.get))
```
